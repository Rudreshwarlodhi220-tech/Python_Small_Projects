# IPL data Analysis Project 

```python
import numpy as np
import pandas as pd
import plotly.express as px
```

```python
data = pd.read_csv('IPL 2022)_P2.csv')
```

```python
data
```

## 1. Number of matches Won by each team in IPL 2022.

```python
fig = px.bar(data , x = 'match_winner' , title = ' Graph Shows how many matches win by each team in IPL 2022',
                                width = 1000, height = 600)
# Yaha y isliye nahi kyuki match_winner wale me se hi count krr lega (kolka , chenai then kolkka...)
# plotly has capability to add automatically add count .

fig.show()
```

```python
match_winner = data['match_winner'].value_counts().reset_index()  # unique krr dega.
match_winner.columns = ['match_winner', 'count']

fig = px.bar(match_winner, x = 'match_winner', y = 'count',
             labels ={'match_winner':'Match Winners', 'count':'Numbers of win'},
            title = 'Matches win by teams',
            color = match_winner[ 'match_winner'] )

fig.show()
```

## 2. Number of match won by chasing and defending.

```python
import plotly.express as px
import pandas as pd

# मान लो तुम्हारे पास dataframe 'data' है
# उसमें 'won_by' नाम का कॉलम है जिसमें values होती हैं: "Chasing" या "Defending"

# Step 1: Count matches won by chasing vs defending
win_counts = data['won_by'].value_counts().reset_index()
win_counts.columns = ['Result', 'Count']

# Step 2: Create pie chart
fig = px.pie(
    win_counts,
    names='Result',        # Chasing / Defending
    values='Count',        # कितने मैच जीते गए
    title='Matches Won by Chasing vs Defending',
    color='Result',        # अलग-अलग रंग
    hole=0.3               # donut-style chart (optional)
)

# Step 3: Show chart
fig.show()
```

```python
win_match_per = data['won_by'].value_counts().reset_index()
win_match_per.columns = ['Result', 'Count'] # 'Wickets','Runs' isliye nahi kyuki ye ak hi column ke hai.

px.pie(win_match_per, names = 'Result', values = 'Count', labels = {'Wickets':'Chasing', 'Runs': 'Defending',},hole = 0.4)
# px.pie me (x ki place pe names ,y = values aata hai)
```
