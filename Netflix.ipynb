```python
df['Date'] = pd.to_datetime(df['Date'])
df = df.set_index('Date')
df.head(5)
```

# Volume of stock traded.

```python
sns.lineplot( x = df.index , y = df['Volume'] , label = 'Volume')
plt.title =(' Volume of stock versus time.')
```

```python
#2 Netflix stock price High, Open, Close.
df.plot(y = ['High','Close','Open'], title = 'Neflix Stoks Price')
```

```python
import matplotlib.pyplot as plt
import pandas as pd

# Assuming df is your DataFrame with stock data
# First, make sure your index is a DatetimeIndex
# If your date column is not the index but exists in the DataFrame as a column (e.g., 'Date')
# df['Date'] = pd.to_datetime(df['Date'])  # Convert to datetime if needed
# df.set_index('Date', inplace=True)       # Set as index

# If you need to convert the existing index to datetime:
# df.index = pd.to_datetime(df.index)

#3 Top-5 Dates with highest stock price.
fig,(ax1,ax2,ax3) = plt.subplots(3,figsize = (15,10))

# Now these will work because the index is a DatetimeIndex
df.groupby(df.index.day).mean().plot(y = 'Volume', ax=ax1, xlabel = 'Day')
df.groupby(df.index.month).mean().plot(y = 'Volume', ax=ax2, xlabel = 'Month')
df.groupby(df.index.year).mean().plot(y = 'Volume', ax=ax3, xlabel = 'Year')

plt.tight_layout()
plt.show()
```

```python
fig, (ax1, ax2, ax3) = plt.subplots(3, figsize=(15,10))

# Group by day, month, year
df.groupby(df.index.day).mean().plot(y='Volume', ax=ax1, xlabel='Day')
df.groupby(df.index.month).mean().plot(y='Volume', ax=ax2, xlabel='Month')
df.groupby(df.index.year).mean().plot(y='Volume', ax=ax3, xlabel='Year')

plt.tight_layout()
plt.show()
```

```python
# 4 Top-5 Dates with highest stock price.

a = df.sort_values(by = 'High',ascending=False).head(5)
a
```

```python
# 5Top five dates with lowest stock price.

b = df.sort_values(by = 'Low',ascending=True).head(5)
b
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Create subplots (1 row, 2 columns)
fig, axes = plt.subplots(nrows=1, ncols=2, sharex=True, figsize=(7,5))

# Add title
fig.suptitle('High and Low values stock per period of time', fontsize=18)

# Plot High values on first axis
sns.lineplot(ax=axes[0], x=df.index, y=df['High'], color='green')

# Plot Low values on second axis
sns.lineplot(ax=axes[1], x=df.index, y=df['Low'], color='red')

plt.tight_layout()
plt.show()
```
